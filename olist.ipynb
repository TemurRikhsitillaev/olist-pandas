```python
import pandas as pd
DATA = 'data/'
orders = pd.read_csv(DATA + 'olist_orders_dataset.csv', parse_dates=[
    'order_purchase_timestamp',
    'order_approved_at',
    'order_delivered_carrier_date',
    'order_delivered_customer_date',
    'order_estimated_delivery_date',
], date_format = {
    'order_purchase_timestamp': '%m/%d/%y %H:%M',
    'order_approved_at': '%m/%d/%y %H:%M',
    'order_delivered_carrier_date': '%m/%d/%y %H:%M',
    'order_delivered_customer_date': '%m/%d/%y %H:%M',
    'order_estimated_delivery_date': '%m/%d/%y %H:%M',
})
```

```python
df = orders.merge(customers, on = 'customer_id')
```

```python
delivered = df[
    (df['order_status'] == 'delivered') &
    (df['order_delivered_customer_date'].notna()) &
    (df['order_purchase_timestamp'] >= '2017-01-01') &
    (df['order_purchase_timestamp'] < '2018-09-01')
    ]
```

```python
delivered.shape
```

```python
assert delivered.shape[0] == 96203
```

```python
purchase_date = delivered['order_purchase_timestamp'].dt.normalize()
delivered_date = delivered['order_delivered_customer_date'].dt.normalize()
estimated_date = delivered['order_estimated_delivery_date'].dt.normalize()

delivered['delivery_days'] = (delivered_date - purchase_date).dt.days
delivered['is_late'] = delivered_date > estimated_date
delivered['days_vs_estimate'] = (delivered_date - estimated_date).dt.days
delivered['purchase_month'] = delivered['order_purchase_timestamp'].dt.to_period('M')
delivered['promise_days'] = (estimated_date - purchase_date).dt.days

delivered
```

```python
assert delivered['delivery_days'].min() == 0
assert delivered['delivery_days'].max() == 210
assert delivered['is_late'].sum() == 6531
assert delivered['delivery_days'].median() == 10
assert delivered['delivery_days'].quantile(0.9) == 23
assert round(delivered['delivery_days'].mean(), 1) == 12.5
```

```python
delivered['bucket'] = pd.cut(
    delivered['delivery_days'],
    bins=[-1,3,7,14,23,30,60,float('inf')],
    labels=['0-3 дн.','4-7 дн.','8-14 дн.','15-23 дн.','24-30 дн.','31-60 дн.', '61+ дн.']
)
```

```python
bucketed = delivered.groupby('bucket').agg(
    orders_count=('order_id', 'size'),
)
bucketed['cum_orders_count'] = bucketed['orders_count'].cumsum()
bucketed['pct'] = (bucketed['orders_count'] / bucketed['orders_count'].sum() * 100).round(2)
bucketed['cum_pct'] = (bucketed['orders_count'].cumsum() / bucketed['orders_count'].sum() * 100).round(2)

bucketed
```

```python
assert bucketed.loc['8-14 дн.', 'orders_count'] == 37899
assert bucketed.loc['15-23 дн.', 'cum_pct'] == 90.57
assert bucketed['orders_count'].sum() == 96203
```

```python
group_by_month = delivered.groupby('purchase_month').agg(
    orders_count=('order_id', 'size'),
    late_orders_count=('is_late', 'sum'),
    median_delivery_days=('delivery_days', 'median'),
    p90_delivery_days=('delivery_days', lambda d: d.quantile(0.9)),
    median_promise_days=('promise_days', 'median'),
)
group_by_month['late_orders_pct'] = (group_by_month['late_orders_count'] / group_by_month['orders_count'] * 100).round(2)
group_by_month['cushion_days'] = (group_by_month['median_promise_days'] - group_by_month['p90_delivery_days']).round(0).astype(int)

group_by_month
```

```python
assert group_by_month['orders_count'].sum() == 96203
assert group_by_month['late_orders_count'].sum() == 6531
assert len(group_by_month) == 20
```

```python
reviews = pd.read_csv(DATA + 'olist_order_reviews_dataset.csv', parse_dates = [
    'review_creation_date', 
    'review_answer_timestamp',
], date_format = {
    'review_creation_date': '%Y-%m-%d %H:%M:%S',
    'review_answer_timestamp': '%Y-%m-%d %H:%M:%S',
}
)
reviews.head()
```

```python
reviews.dtypes
```

```python
# orders with more than one review
(reviews['order_id'].value_counts() > 1).sum()
```

```python
reviews['order_id'].nunique()
```

```python
# keep lastest review of orders
reviews_by_order = reviews.sort_values(['order_id', 'review_creation_date', 'review_id'], ascending=[True, False, False]).drop_duplicates(subset='order_id', keep='first')
```

```python
orders_reviews = delivered.merge(reviews_by_order, on='order_id', how='left')

orders_reviews.info()
```

```python
group_by_review_score = orders_reviews.groupby('review_score', dropna=False).agg(orders_count=('order_id', 'count'))
group_by_review_score['pct'] = (group_by_review_score['orders_count'] / group_by_review_score['orders_count'].sum() * 100).round(2)

group_by_review_score
```

```python
orders_reviews_notna = delivered.merge(reviews_by_order, on='order_id', how='inner')

orders_reviews_notna.info()
```

```python
group_orders_reviews_notna = orders_reviews_notna.groupby('is_late').agg(
    orders_count=('order_id', 'count'),
    avg_review_score=('review_score', 'mean'),
    median_review_score=('review_score', 'median'),
    orders_with_score_1=('review_score', lambda a: (a == 1).sum()),
    orders_with_score_2=('review_score', lambda a: (a == 2).sum()),
)

group_orders_reviews_notna['pct_orders_with_score_1'] = (
    group_orders_reviews_notna['orders_with_score_1'] / group_orders_reviews_notna['orders_count'] * 100
).round(2)

group_orders_reviews_notna['pct_orders_with_score_2'] = (
    group_orders_reviews_notna['orders_with_score_2'] / group_orders_reviews_notna['orders_count'] * 100
).round(2)

group_orders_reviews_notna['avg_review_score'] = group_orders_reviews_notna['avg_review_score'].round(2)
group_orders_reviews_notna['median_review_score'] = group_orders_reviews_notna['median_review_score'].round(0).astype('int')

group_orders_reviews_notna
```
